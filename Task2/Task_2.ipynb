# Q1. Find the Shortest Word in a Sentence
- Write a Python program to find the shortest word in a given sentence.
- Example input:
- Python makes coding easy
- Expected output:
- easy


```python
# take the sentence as str
# split using space as delimitter
# using for loop find each word length
# check which is having less length
```

```python
sentence='Python makes coding easy'
def short_word(sentence):
    if not sentence:
        return 0
    splitted=sentence.split(" ")
    s=splitted[0]
    for i in splitted:
        if len(i)<len(s):
            s=i
    return s
short_word(sentence)
```

```python
sen2="Innomatics is having good Data Science Course"
short_word(sen2)
```

## Q6. Count Vowels in a String
- Write a Python program to count the total number of vowels in a string. Consider both uppercase and lowercase vowels.
- Example input:
- Education
- Expected output:
- 5


```python
# take the sentence and add a counter for the vowels to count
# using a loop iterate through the word characters and using a if count the vowels 
```

```python
def count_vowel(word):
    count=0
    for i in word.lower():
        if i.isalpha(): 
            if i in 'aeiou':
                count+=1
    return count
word1="Education"
count_vowel(word1)
```

## Q8. Find the Most Frequent Character
- Write a Python program to find the character that appears most frequently in a string. If multiple characters have the same highest frequency, - - --- return the first one encountered. Ignore spaces.
- Example input:
- banana
- Expected output:
- a


```python
# take a dictionary
# replace the space characters in the word string and loop through the word
# using if condition update the key-val pairs as count of each char in the word
# find the max value in the values in dictionary 
# using for loop iterate through the items in the dictionary
# find the maximum match value in the dictionary and return the key as output
```

```python
def most_freq(word):
    d={}
    words=word.replace(" ","")
    for i in words:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    maxim=max(d.values())
    for i, j in d.items():
        if j==maxim:
            return i
most_freq("banana")
```

```python
most_freq("Innomatics")
```

- Write a Python program to find the character that appears least frequently in a string. If multiple characters have the same lowest frequency, return - the first one encountered. Ignore spaces.
- Example input:

- Expected output:

```python
# take a dictionary
# replace the space characters in the word string and loop through the word
# using if condition update the key-val pairs as count of each char in the word
# find the min value in the values in dictionary 
# using for loop iterate through the items in the dictionary
# find the minimum match value in the dictionary and return the key as output
```

```python
def least_freq(word):
    d={}
    words=word.replace(" ","")
    for i in words:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    minim=min(d.values())
    for i, j in d.items():
        if j==minim:
            return i
least_freq("sucess")
```

## Q10. Find All Characters Appearing More Than Once

- Example input:
- programming
- Expected output:
- ['r', 'g', 'm']


```python
# remove all the space characters using replace method
# create a dictionary and add the key val pair as frequency of each char
# appending the keys which have the values greater than one
```

```python
def char_more(word):
    no_spaces=word.replace(" ","")
    d={}
    for i in no_spaces:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    a=[]
    for i,j in d.items():
        if j>1:
            a.append(i)
    return a
char_more("programming")
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status
- Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.
- Example input:
- [1, 2, 3, 4, 5]
- Expected output:
- {1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}


```python
# create a dictionay and using a for loop iterate through the list given
# check each num as even or odd by dividing the number by 2
# update the key as number and even or odd as value as value in the dictionary
```

```python
def status(nums):
    d={}
    for i in nums:
        if i%2==0:
            d[i]="Even"
        else:
            d[i]="Odd"
    return d
status([5,4,7,8,10])
```

## Q16. Find the Most Frequent Word in a Sentence
- Write a Python program to find the word that appears most frequently in a sentence. Treat uppercase and lowercase versions of a word as the same -  - word. If multiple words share the highest frequency, return the one that appears first.
- Example input:
- Python is easy and Python is powerful
- Expected output:
- python


```python
# firstly split the sentence so that spaces can be ignored 
# create a dictionary with frequency of each word
# find the maximum frequency word in the sentence
# loop through the dctionary and find the frequency that matches maximum frequency and extract the word
```

```python
def most_freq_word(sentence):
    splitted=sentence.split(" ")
    d={}
    for i in splitted:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    most_freq=max(d.values())
    for i,j in d.items():
        if j==most_freq:
            return i
most_freq_word("Python is easy and Python is powerful")
```

## Q20. Find the Longest Word That Appears Only Once
- Write a Python program to find the longest word that appears exactly once in a sentence. If multiple qualifying words have the same maximum length, - return the first one encountered. Treat uppercase and lowercase versions as the same word.
- Example input:
- Python makes coding useful Python
- Expected output:
- coding


```python
# firstly split the sentence so that spaces can be ignored 
# create a dictionary with frequency of each word
# create a new list
# loop through the dictionary and find the value that matches exactly one and append the words into the list
# find the maximum length word in the list 
```

```python
def long_word_once(sentence):
    splitted=sentence.lower().split(" ")
    d={}
    for i in splitted:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    a=[]
    for i,j in d.items():
        if j==1:
            a.append(i)
    return max(a,key=len)
long_word_once("Python makes coding useful Python")
```
