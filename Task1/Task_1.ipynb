## Q35

## Q36

```python
# a is a list and b is deepcopy of a which creates a new memory loc 
# updating b[0][1]=99, it will change in b only so a will be [[1, 2], [3, 4]]
```

```python
import copy

a = [[1, 2], [3, 4]]
b = copy.deepcopy(a)
b[0][1] = 99
print(a)
```
